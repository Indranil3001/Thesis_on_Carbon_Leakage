```python

C_TO_CO2 = 3.664


def to_long_format(df, value_name):
    
    return df.melt(id_vars='Year', var_name='Country', value_name=value_name)


def carbon_to_co2(df, value_name):
    
    df = df.copy()
    df[value_name] = df[value_name] * C_TO_CO2
    return df


df_c_long = carbon_to_co2(to_long_format(df_c, 'Consumption_Emissions'), 'Consumption_Emissions')
df_p_long = carbon_to_co2(to_long_format(df_p, 'Production_Emissions'), 'Production_Emissions')
```


**On the factor:** GCP reports Mt of *carbon*; this thesis reports Mt of *CO₂*.
The stoichiometric ratio is 44/12 = 3.6667, but GCB publishes 3.664, so 3.664 is
kept to stay consistent with the source. Every emissions figure downstream —
`transfer` included — is therefore in **Mt CO₂**.

# **Treatment Block** 

```python
ANNEX_B = {
    'AUS','AUT','BEL','BGR','CAN','HRV','CZE','DNK','EST','FIN','FRA','DEU',
    'GRC','HUN','ISL','IRL','ITA','JPN','LVA','LIE','LTU','LUX','MCO','NLD',
    'NZL','NOR','POL','PRT','ROU','RUS','SVK','SVN','ESP','SWE','CHE','UKR','GBR',
}  # USA excluded: as theysigned but never-treated.

# Catch typos/format drift instead of silently coding a treated country as control
missing = ANNEX_B - set(df_final['ISO3'])
assert missing <= {'ISL','LIE','MCO'}, f"unexpected Annex B codes absent: {missing}"

KYOTO_ENTRY_INTO_FORCE  = 2005   # treaty legally binding from 16 Feb 2005
FIRST_COMMITMENT_PERIOD = 2008   # 2008-2012 compliance window

df_final['annex_b']     = df_final['ISO3'].isin(ANNEX_B).astype(int)
df_final['first_treat'] = df_final['annex_b'] * KYOTO_ENTRY_INTO_FORCE  # 0 = never-treated
df_final['post']        = (df_final['Year'] >= KYOTO_ENTRY_INTO_FORCE).astype(int)
df_final['rel_year']    = np.where(df_final['annex_b'] == 1,
                                   df_final['Year'] - KYOTO_ENTRY_INTO_FORCE, np.nan)  # event-study axis

# Same coding anchored on 2008, for the timing robustness check.
df_final['first_treat_2008'] = df_final['annex_b'] * FIRST_COMMITMENT_PERIOD
df_final['post_2008']        = (df_final['Year'] >= FIRST_COMMITMENT_PERIOD).astype(int)
df_final['rel_year_2008']    = np.where(df_final['annex_b'] == 1,
                                        df_final['Year'] - FIRST_COMMITMENT_PERIOD, np.nan)

# C&S needs a numeric unit id
df_final['country_id'] = df_final['ISO3'].map(
    {iso: i for i, iso in enumerate(sorted(df_final['ISO3'].unique()))}
)

print(df_final.groupby('annex_b')['ISO3'].nunique())         
print(df_final.groupby('first_treat')['ISO3'].nunique())     
```

**Treatment coding for the DiD.**

`ANNEX_B` is the set of countries with binding first-period Kyoto targets. The USA
is deliberately excluded: it signed but never ratified, so it never faced a binding
target and belongs in the never-treated control group. The assertion guards against
a code silently dropping out of the panel — `ISL`, `LIE` and `MCO` are allowed to be
absent because they are too small to appear in all source files.

**On the timing.** Two dates are defensible and the choice is a real modelling
decision, not a detail:

- **2005** — the treaty entered into force on 16 Feb 2005, so this is when the obligation became binding and anticipation effects could start.
- **2008** — the first commitment period (2008–2012) is when compliance was actually measured.

2005 is the main specification; `first_treat_2008`, `post_2008` and `rel_year_2008`
carry the alternative so the robustness check is a one-line change at estimation
rather than a re-run of this notebook.

**A note on the estimator.** `first_treat` follows the Callaway–Sant'Anna
convention (`0` = never-treated). But every treated unit here shares a single
adoption date, so there is no staggered adoption — C&S collapses to essentially a
2×2 design and buys little over a two-way fixed effects model with an event-study
plot on `rel_year`. Both should give the same answer; TWFE is the simpler thing to
defend in the write-up.
